接下来会是使用类来描述时钟

```python
"""
    这是一个定义类描写时钟的程序
"""
import time
class Clock:
    def __init__(self,hour = 0,minute = 0,second = 0):
        self.hour = hour
        self.minute = minute
        self.second = second

    def clock_run(self):
        self.second += 1
        if self.second == 60:
            self.second = 0
            self.minute += 1
            if self.minute == 60:
                self.minute = 0
                self.hour += 1
                if self.hour == 24:
                    self.hour = 0
    def show_time(self):
        print(f"{self.hour:02d}:{self.minute:02d}:{self.second:02d}")

clock = Clock(23,59,40)

while True:
    clock.show_time()
    time.sleep(1)
    clock.clock_run()
    if clock.hour == 0 and clock.minute == 0 and clock.second == 0:
        break
```

下面是使用类来描述平面上的点，同时能够计算平面两个点之间的距离

```python
class Point:
    """平面上的点"""

    def __init__(self, x=0, y=0):
        """初始化方法
        :param x: 横坐标
        :param y: 纵坐标
        """
        self.x, self.y = x, y

    def distance_to(self, other):
        """计算与另一个点的距离
        :param other: 另一个点
        """
        dx = self.x - other.x
        dy = self.y - other.y
        return (dx * dx + dy * dy) ** 0.5

    def __str__(self):
        return f'({self.x}, {self.y})'


p1 = Point(3, 39)
p2 = Point(6, 91)
print(p1)  # 调用对象的__str__魔法方法
print(p2)
print(p1.distance_to(p2))
```

对于python这种语言是动态语言，在运行时，也能够改变他的结构，比如引入新的属性，对象或者是函数。下面的代码就是在对象中引入了一个新的属性.如果不想对象引入新的属性，可以使用__slots__。

```python
class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age


stu = Student('王大锤', 20)
stu.sex = '男'  # 给学生对象动态添加sex属性
print(stu.name)  # 王大锤
print(stu.age)   # 20
print(stu.sex)   # 男

class Student:
    __slots__ = ('name','age')          #指定了name,age参数，如果再给对象添加其他属性会报错
    def __init__(self, name, age):
        self.name = name
        self.age = age


stu = Student('王大锤', 20)
stu.sex = '男'  # 给学生对象动态添加sex属性
```

类中定义的方法都是对象方法，但是有时候这个对象并不能成立，所以除了对象方法以外还需要类方法以及静态方法，静态方法就是只需要参数，不需要对象，用例子中的代码来说就是能够去判断这个对象在生成前是否成立。对于类方法，就是能够调用类本身来做一些事

```python
from math import sqrt

class Triangle:
    """三角形类"""

    def __init__(self,a,b,c):
        if not Triangle.is_valid(a,b,c):
            raise ValueError(f"不能构成三角形的三条边长: {a}, {b}, {c}")
        self.a = a
        self.b = b
        self.c = c

    #实例方法：读取对象边长计算三角形的周长
    def perimeter(self):
        return self.a + self.b + self.c
    #静态方法：只传入三个数字
    @staticmethod
    def is_valid(a,b,c):
        return a + b > c and a + c > b and b + c > a
    #类方法：传入类对象cls和三个数字
    @classmethod
    def create(cls,a,b,c):
        if not cls.is_valid(a,b,c):
            raise ValueError(f"不能构成三角形的三条边长: {a}, {b}, {c}")
        return cls(a,b,c)
t = Triangle(3, 4, 5)

print(t.perimeter())  # 12
print(Triangle.is_valid(3, 4, 5))  # True
print(Triangle.is_valid(1, 2, 3))  # False

t2 = Triangle.create(3, 4, 5)
print(t2.perimeter())  # 12
```

在类的里面蒸架一个property装饰器可以将方法调用改为属性，观察下面两个代码，一个是方法调用t.perimeter(),另一个是t.perimeter,读取属性时自动计算了

```python
class Triangle(object):
    """三角形"""

    def __init__(self, a, b, c):
        """初始化方法"""
        self.a = a
        self.b = b
        self.c = c

    @staticmethod
    def is_valid(a, b, c):
        """判断三条边长能否构成三角形(静态方法)"""
        return a + b > c and b + c > a and a + c > b

    # @classmethod
    # def is_valid(cls, a, b, c):
    #     """判断三条边长能否构成三角形(类方法)"""
    #     return a + b > c and b + c > a and a + c > b

    def perimeter(self):
        """计算周长"""
        return self.a + self.b + self.c

    def area(self):
        """计算面积"""
        p = self.perimeter() / 2
        return (p * (p - self.a) * (p - self.b) * (p - self.c)) ** 0.5
      

if Triangle.is_valid(3, 4, 5):
    t = Triangle(3, 4, 5)
    print(f'周长: {t.perimeter()}')
    print(f'面积: {t.area()}')
else:
    print('无效的边长!!!')
```

```python
class Triangle(object):#object是所有类的父类
    """三角形"""

    def __init__(self, a, b, c):
        """初始化方法"""
        self.a = a
        self.b = b
        self.c = c

    @staticmethod
    def is_valid(a, b, c):
        """判断三条边长能否构成三角形(静态方法)"""
        return a + b > c and b + c > a and a + c > b

    @property
    def perimeter(self):
        """计算周长"""
        return self.a + self.b + self.c

    @property
    def area(self):
        """计算面积"""
        p = self.perimeter / 2
        return (p * (p - self.a) * (p - self.b) * (p - self.c)) ** 0.5


if Triangle.is_valid(3, 4, 5):
    t = Triangle(3, 4, 5)
    print(f'周长: {t.perimeter}')
    print(f'面积: {t.area}')
else:
    print('无效的边长!!!')
```

接下来，就是继承以及多态的问题，定义类的时候是可以继承父类基础功能，比如上面的Triangle类继承的是object类，这个object类是所有类的父类。下面就用person类和student类以及teacher类举例。子类是student以及teacher类都继承了person类的基础特性，吃饭睡觉，而子类之间亦有差距。然后类的初始化可以super().__init__(name, age)调用父类的初始化，实现初始化的公共部分，减少代码

```python
class Person:
    """人"""

    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    def eat(self):
        print(f'{self.name}正在吃饭.')
    
    def sleep(self):
        print(f'{self.name}正在睡觉.')


class Student(Person):
    """学生"""
    
    def __init__(self, name, age):
        super().__init__(name, age)
    
    def study(self, course_name):
        print(f'{self.name}正在学习{course_name}.')


class Teacher(Person):
    """老师"""

    def __init__(self, name, age, title):
        super().__init__(name, age)
        self.title = title
    
    def teach(self, course_name):
        print(f'{self.name}{self.title}正在讲授{course_name}.')



stu1 = Student('白元芳', 21)
stu2 = Student('狄仁杰', 22)
tea1 = Teacher('武则天', 35, '副教授')
stu1.eat()
stu2.sleep()
tea1.eat()
stu1.study('Python程序设计')
tea1.teach('Python程序设计')
stu2.study('数据科学导论')
```
