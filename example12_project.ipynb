```python
"""
    类与对象
    这一程序主要是关于类与对象。
"""
#创建类并且初始化

class Student:
    def __init__(self,name,age):
        self.name = name
        self.age = age

    def name_age(self):
        print(f"{self.name}, {self.age} years old.")

    def study(self):
        print(f"{self.name} is studying.")


stu1 = Student("czq",22)
stu2 = Student("lx",21)
stu1.name_age()
stu2.name_age()
stu1.study()
stu2.study()
```

接下来会是使用类来描述时钟

```python
"""
    这是一个定义类描写时钟的程序
"""
import time
class Clock:
    def __init__(self,hour = 0,minute = 0,second = 0):
        self.hour = hour
        self.minute = minute
        self.second = second

    def clock_run(self):
        self.second += 1
        if self.second == 60:
            self.second = 0
            self.minute += 1
            if self.minute == 60:
                self.minute = 0
                self.hour += 1
                if self.hour == 24:
                    self.hour = 0
    def show_time(self):
        print(f"{self.hour:02d}:{self.minute:02d}:{self.second:02d}")

clock = Clock(23,59,40)

while True:
    clock.show_time()
    time.sleep(1)
    clock.clock_run()
    if clock.hour == 0 and clock.minute == 0 and clock.second == 0:
        break
```

下面是使用类来描述平面上的点，同时能够计算平面两个点之间的距离

```python
class Point:
    """平面上的点"""

    def __init__(self, x=0, y=0):
        """初始化方法
        :param x: 横坐标
        :param y: 纵坐标
        """
        self.x, self.y = x, y

    def distance_to(self, other):
        """计算与另一个点的距离
        :param other: 另一个点
        """
        dx = self.x - other.x
        dy = self.y - other.y
        return (dx * dx + dy * dy) ** 0.5

    def __str__(self):
        return f'({self.x}, {self.y})'


p1 = Point(3, 39)
p2 = Point(6, 91)
print(p1)  # 调用对象的__str__魔法方法
print(p2)
print(p1.distance_to(p2))
```

对于上面时钟还是学生或者是点对象来说，其中都有一些参数（属性），这些属性有的是需要private，有的是需要protected,有的就需要public。一般在属性前__实现私有属性（__name）,用‘_’实现保护属性（_name），而通常都是public的属性。运行下面的程序会发现，会出现AttributeErorr.
因为属性的私有化，无法在类外访问该属性，但是可以通过stu._Student__name访问，这是因为python语言的开放性，能够去访问

```python
class Student:

    def __init__(self, name, age):
        self.__name = name
        self.__age = age

    def study(self, course_name):
        print(f'{self.__name}正在学习{course_name}.')


stu = Student('王大锤', 20)
stu.study('Python程序设计')
print(stu._Student__name)  # 王大锤
print(stu.__name)  # AttributeError: 'Student' object has no attribute '__name'
```

对于python这种语言是动态语言，在运行时，也能够改变他的结构，比如引入新的属性，对象或者是函数。下面的代码就是在对象中引入了一个新的属性.如果不想对象引入新的属性，可以使用__slots__。

```python
class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age


stu = Student('王大锤', 20)
stu.sex = '男'  # 给学生对象动态添加sex属性
print(stu.name)  # 王大锤
print(stu.age)   # 20
print(stu.sex)   # 男
class Student:
    __slots__ = ('name','age')
    def __init__(self, name, age):
        self.name = name
        self.age = age


stu = Student('王大锤', 20)
stu.sex = '男'  # 给学生对象动态添加sex属性
```
