读取文件讲完了，现在就是写入和追加，即”w“和”a“,使用w会成功打开文件并且清除文件内容，文件不存在创建文件。如果使用a就是在文件末尾追加，如果没有文件创建文件。现在下方代码实现。

```python
path =  path = '../mnist_cnn/沁园春·雪.md'
with open(path, 'w', encoding='utf-8') as f:
    f.write('沁园春·雪\n')
    f.write('作者：毛泽东\n')
    f.write('北国风光，千里冰封，万里雪飘。\n')
    f.write('望长城内外，惟余莽莽；大河上下，顿失滔滔。\n')
    f.write('山舞银蛇，原驰蜡象，欲与天公试比高。\n')
    f.write('须晴日，看红装素裹，分外妖娆。\n')
    f.write('江山如此多娇，引无数英雄竞折腰。\n')
    f.write('惜秦皇汉武，略输文采；唐宗宋祖，稍逊风骚。\n')
    f.write('一代天骄，成吉思汗，只识弯弓射大雕。\n')
    f.write('俱往矣，数风流人物，还看今朝。')
with open(path, 'r', encoding='utf-8') as f:
    print(f.read()) 
```

文件的打开读写都学习了，但是会存在很多的报错情况，所以需要异常处理，这里用到的关键字是try，expect,finally

```python
path  = '../mnist_cnn/沁园春·雪.md'
try:
    with open(path, 'r', encoding='utf-8') as f:
        text = f.read()

except FileNotFoundError:
    print(f"文件 {path} 未找到。")
except IOError:
    print(f"读取文件 {path} 时发生错误。") 
except UnicodeDecodeError:
    print(f"解码文件 {path} 时发生错误。")
finally:
    print("文件操作完成。")
    print(text)
```
